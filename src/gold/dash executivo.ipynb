```python
%sql

SELECT *,
        SUM(nrQuantidadeTransacoes) OVER (ORDER BY dtRef) AS nrQuantidadeTransacoesAcum,
        SUM(nrQuantidadePontos) OVER (ORDER BY dtRef) AS nrQuantidadePontosAcum,
        SUM(nrQuantidadePontosPos) OVER (ORDER BY dtRef) AS nrQuantidadePontosAcumPos,
        (-1) * SUM(nrQuantidadePontosNeg) OVER (ORDER BY dtRef) AS nrQuantidadePontosAcumNeg
FROM gold.upsell.daily_report 
WHERE dtRef IS NOT NULL AND descNomeProduto IS NULL AND nrQuantidadePontosNeg > -70000
ORDER BY dtRef DESC
```

```python
%sql

SELECT  dtRef, 
        any_value(IdProduto), 
        any_value(descNomeProduto), 
        sum(nrQuantidadeTransacoes), 
        sum(nrQuantidadeClientes), 
        sum(nrQuantidadePontos), 
        sum(nrQuantidadePontosPos), 
        sum(nrQuantidadePontosNeg)

FROM gold.upsell.monthly_report 

WHERE dtRef IS NOT NULL AND IdProduto IS NULL AND nrQuantidadePontos > -50000

GROUP BY dtRef
ORDER BY dtRef DESC
```

```python
%sql
SELECT YEAR(dtRef) as yrRef,  
        descNomeProduto,
        sum(nrQuantidadeTransacoes),
        sum(nrQuantidadeClientes),
        sum(nrQuantidadePontos),
        sum(nrQuantidadePontosPos),
        sum(nrQuantidadePontosNeg)


FROM gold.upsell.daily_report 

WHERE descNomeProduto IS NOT NULL 

GROUP BY yrRef, descNomeProduto

ORDER BY yrRef DESC
```

```python
%sql

WITH produtos AS (
    SELECT
        descNomeProduto,
        SUM(nrQuantidadeTransacoes) AS transacoes,
        SUM(nrQuantidadeClientes) AS clientes,
        SUM(nrQuantidadePontos) AS pontos,
        SUM(nrQuantidadePontosPos) AS pontosPos,
        SUM(nrQuantidadePontosNeg) AS pontosNeg
    FROM gold.upsell.daily_report
    WHERE descNomeProduto IS NOT NULL
    GROUP BY descNomeProduto
),

ranking AS (
    SELECT
        *,
        ROW_NUMBER() OVER (ORDER BY transacoes DESC) AS rn
    FROM produtos
)

SELECT
    CASE
        WHEN rn <= 5 THEN descNomeProduto
        ELSE 'Outros'
    END AS descNomeProduto,

    SUM(transacoes) AS transacoes,
    SUM(clientes) AS clientes,
    SUM(pontos) AS pontos,
    SUM(pontosPos) AS pontosPos,
    SUM(pontosNeg) AS pontosNeg

FROM ranking
GROUP BY
    CASE
        WHEN rn <= 5 THEN descNomeProduto
        ELSE 'Outros'
    END

ORDER BY transacoes DESC;
```
